```python
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
plt.figure(figsize=(10, 10))
sns.boxplot(data=df, x="accommodation_type",
            y="total_trip_spend_usd")
plt.title("outlier detection for total trip spending")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

outlier found in total_trip_spend_usd

```python

Q1 = df["total_trip_spend_usd"].quantile(0.25)
print(Q1)
```

```python
Q3 = df["total_trip_spend_usd"].quantile(0.75)
print(Q3)
```

```python
IQR = Q3 - Q1
print(IQR)
```

```python
lower_limit = Q1 - 1.5 * IQR
print(lower_limit)
```

```python
upper_limit = Q3 + 1.5 * IQR
print(upper_limit)
```

```python
df[
    (df["total_trip_spend_usd"] < lower_limit) |
    (df["total_trip_spend_usd"] > upper_limit)
]
```

neglecting the outlier in the total money spend as they may be genuine travel pattern

12.HEAT MAP FINDING CORRELATION BETWEEN NUMERIC VARIABLES

```python
columns = [
    "num_travelers",
    "duration_nights",
    "advance_booking_days",
    "budget_per_person_usd",
    "total_trip_spend_usd",
    "restaurant_spend_per_day_usd",
    "attractions_visited",
    "trip_rating",
    "carbon_footprint_kg_co2"
]
corr_matrix = df[columns].corr()
fig = px.imshow(
    corr_matrix,
    text_auto=".2f",
    color_continuous_scale="RdBu_r",
    zmin=-1,
    zmax=1,
    aspect="auto",
    title="Correlation Heatmap of Tourism Data"
)
fig.update_layout(
    width=1000,
    height=800,
    xaxis_title="Numerical Variables",
    yaxis_title="Numerical Variables"
)

fig.show()
```
